```python
#3. Default Arguments
#create table emp(id int,name varchar(20),exp int default 0);
#insert into emp values(1,'irfan',23);
#insert into emp values(2,'sathya');

def div(x,y=1):#Default the param values if some default value has to be substituted in the runtime
#Note: Default parameters has to be after regular parameters
   z=x/y
   return z
res=div(x=10,y=20)#y is overriding 1 with 20
print(res)
res=div(10)#y is considered as default 1 (positional + default)
print(res)
res=div(x=10)
print(res)#y is considered as default 1 (named + default)
```

```python
def calc_sal(**kwargs):
    incentive_amt=kwargs.get('sal',0)*kwargs.get('inc',0)
    final_sal_to_deposit=kwargs.get('sal',0)+incentive_amt+kwargs.get('bon',0)
    return kwargs.get('comp'),final_sal_to_deposit

print(calc_sal(comp='hcl',sal=100000))#100000+(100000*.3)=100000
print(calc_sal(comp='infy',sal=100000,inc=.03))#100000+(100000*.3)=100000
print(calc_sal(comp='cts',sal=100000,inc=.03,bon=5000))
print(calc_sal(bon=10000,comp='inceptez'))
```

```python
#Not adviced to go with positional, since we are passing & manipulating arguments based on name/keywords, hence arbitrary keyword arg is the better approach to follow
def calc_sal(*args):
    company=args[0]
    if company=='hcl':
        salary=args[1]
        incentive=0
        bonus=0
    elif company=='infy':
        salary=args[1]
        incentive=args[2]
        bonus=0
    elif company=='inceptez':
        salary=0
        incentive=0
        bonus=args[3]
    elif company=='cts':
        salary=args[1]
        incentive=args[2]
        bonus=args[3]
    incentive_amt=salary*incentive
    final_sal_to_deposit=salary+incentive_amt+bonus
    return company,final_sal_to_deposit

print(calc_sal('hcl',100000))#100000+(100000*.3)=100000
print(calc_sal('infy',100000,.03))#100000+(100000*.3)=100000
print(calc_sal('cts',100000,.03,5000))
print(calc_sal('inceptez',None,None,10000))
```
