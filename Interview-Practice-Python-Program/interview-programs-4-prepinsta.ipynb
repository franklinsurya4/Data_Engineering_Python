```python
# 101. Check if a string contains only digits

#Method 1: Using isdigit()

s = input("Enter a string: ")

if s.isdigit():
    print("String contains only digits")
else:
    print("String does not contain only digits")
```

```python
#Method 2: Without using built-in methods

s = input("Enter a string: ")
flag = True

if len(s) == 0:
    flag = False
else:
    for ch in s:
        if ch < '0' or ch > '9':
            flag = False
            break

if flag:
    print("String contains only digits")
else:
    print("String does not contain only digits")
```

```python
#102. Rotate an array left by K positions using slicing

arr = list(map(int, input("enter a elements").split()))

k = int(input("enter k elements"))

n = len(arr)
k = k % n

rotated_array = arr[k:] + arr[:k]

print("rotated array is ", rotated_array)
```

```python
#103. Find the missing number in an array (1 to N)

arr = list(map(int, input("Enter array elements: ").split()))

n = len(arr) + 1

total = n * (n + 1) // 2

arr_sum = sum(arr)

missing = total - arr_sum

print("Missing number is:", missing)
```

```python
#104. Find duplicate elements in an array

arr = list(map(int, input("Enter array elements: ").split()))

duplicates = []

for i in range(len(arr)):
    for j in range(i + 1, len(arr)):
        if arr[i] == arr[j] and arr[i] not in duplicates:
            duplicates.append(arr[i])

print("Duplicate elements are:", duplicates)
```

```python
#105. Sum of diagonal elements of a matrix

n = int(input("Enter matrix size: "))

matrix = []

for i in range(n):
    row = list(map(int, input("Enter row elements: ").split()))
    matrix.append(row)

total = 0

for i in range(n):
    total += matrix[i][i]

print("Sum of diagonal elements:", total)
```

```python
# 106. Matrix multiplication

r1 = int(input("Enter rows of first matrix: "))
c1 = int(input("Enter columns of first matrix: "))

A = []

for i in range(r1):
    row = list(map(int, input("Enter row elements: ").split()))
    A.append(row)

r2 = int(input("Enter rows of second matrix: "))
c2 = int(input("Enter columns of second matrix: "))

B = []

for i in range(r2):
    row = list(map(int, input("Enter row elements: ").split()))
    B.append(row)

if c1 != r2:
    print("Matrix multiplication is not possible")
else:
    result = [[0 for j in range(c2)] for i in range(r1)]

    for i in range(r1):
        for j in range(c2):
            for k in range(c1):
                result[i][j] += A[i][k] * B[k][j]

    print("Result matrix:")
    for row in result:
        print(row)
```

```python
#107. Symmetric Matrix Check

'''
This matrix is symmetric because:
- A[0][1] = A[1][0] → 2 = 2
- A[0][2] = A[2][0] → 3 = 3
- A[1][2] = A[2][1] → 5 = 5
'''

n=int(input("Enter a matrix size:"))

matrix=[]

for i in range(n):
    row=list(map(int,input("enter a row elements:").split()))
    matrix.append(row)

symmetric=True

for i in range(n): #rows
    for j in range(n): #columns
        if matrix[i][j] != matrix[j][i]:
            symmetric=False
            break
    if not symmetric:
        break

if symmetric:
    print("symmetric")
else:
    print("not symmetric")
```

```python
#108. Identity Matrix Check

'''
An identity matrix is a square matrix in which:

- All diagonal elements are 1.
- All non-diagonal elements are 0.
'''

n = int(input("Enter matrix size: "))

matrix = []

for i in range(n):
    row = list(map(int, input("Enter row elements: ").split()))
    matrix.append(row)

identity = True

for i in range(n): #rows
    for j in range(n): #columns
        if i == j: # i represents the row index 0, and j represents the column index 0.
            if matrix[i][j] != 1:
                identity = False
                break
        else:
            if matrix[i][j] != 0:
                identity = False
                break

    if not identity:
        break

if identity:
    print("Identity matrix")
else:
    print("Not an identity matrix")
```

```python
# 109. Row-wise and column-wise sum

r = int(input("Enter number of rows: "))
c = int(input("Enter number of columns: "))

matrix = []

for i in range(r):
    row = list(map(int, input("Enter row elements: ").split()))
    matrix.append(row)

# Row-wise sum
for i in range(r):
    total = 0

    for j in range(c):
        total += matrix[i][j]

    print("Row", i + 1, "sum:", total)

# Column-wise sum
for j in range(c):
    total = 0

    for i in range(r):
        total += matrix[i][j]

    print("Column", j + 1, "sum:", total)
```

```python
# 110. Maximum and minimum element in a matrix

r = int(input("Enter number of rows: "))
c = int(input("Enter number of columns: "))

matrix = []

for i in range(r):
    row = list(map(int, input("Enter row elements: ").split()))
    matrix.append(row)

maximum = matrix[0][0]
minimum = matrix[0][0]

for i in range(r):
    for j in range(c):
        if matrix[i][j] > maximum:
            maximum = matrix[i][j]

        if matrix[i][j] < minimum:
            minimum = matrix[i][j]

print("Maximum element:", maximum)
print("Minimum element:", minimum)
```
