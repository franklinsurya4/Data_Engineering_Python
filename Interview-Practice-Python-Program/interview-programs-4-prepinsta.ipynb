```python
# 101. Check if a string contains only digits

#Method 1: Using isdigit()

s = input("Enter a string: ")

if s.isdigit():
    print("String contains only digits")
else:
    print("String does not contain only digits")
```

```python
#Method 2: Without using built-in methods

s = input("Enter a string: ")
flag = True

if len(s) == 0:
    flag = False
else:
    for ch in s:
        if ch < '0' or ch > '9':
            flag = False
            break

if flag:
    print("String contains only digits")
else:
    print("String does not contain only digits")
```

```python
#103. Find the missing number in an array (1 to N)

arr = list(map(int, input("Enter array elements: ").split()))

n = len(arr) + 1

total = n * (n + 1) // 2

arr_sum = sum(arr)

missing = total - arr_sum

print("Missing number is:", missing)
```

```python
#104. Find duplicate elements in an array

arr = list(map(int, input("Enter array elements: ").split()))

duplicates = []

for i in range(len(arr)):
    for j in range(i + 1, len(arr)):
        if arr[i] == arr[j] and arr[i] not in duplicates:
            duplicates.append(arr[i])

print("Duplicate elements are:", duplicates)
```

```python
#105. Sum of diagonal elements of a matrix

n = int(input("Enter matrix size: "))

matrix = []

for i in range(n):
    row = list(map(int, input("Enter row elements: ").split()))
    matrix.append(row)

total = 0

for i in range(n):
    total += matrix[i][i]

print("Sum of diagonal elements:", total)
```

```python
# 106. Matrix multiplication

r1 = int(input("Enter rows of first matrix: "))
c1 = int(input("Enter columns of first matrix: "))

A = []

for i in range(r1):
    row = list(map(int, input("Enter row elements: ").split()))
    A.append(row)

r2 = int(input("Enter rows of second matrix: "))
c2 = int(input("Enter columns of second matrix: "))

B = []

for i in range(r2):
    row = list(map(int, input("Enter row elements: ").split()))
    B.append(row)

if c1 != r2:
    print("Matrix multiplication is not possible")
else:
    result = [[0 for j in range(c2)] for i in range(r1)]

    for i in range(r1):
        for j in range(c2):
            for k in range(c1):
                result[i][j] += A[i][k] * B[k][j]

    print("Result matrix:")
    for row in result:
        print(row)
```

```python
#107. Symmetric Matrix Check

'''
This matrix is symmetric because:
- A[0][1] = A[1][0] → 2 = 2
- A[0][2] = A[2][0] → 3 = 3
- A[1][2] = A[2][1] → 5 = 5
'''

n=int(input("Enter a matrix size:"))

matrix=[]

for i in range(n):
    row=list(map(int,input("enter a row elements:").split()))
    matrix.append(row)

symmetric=True

for i in range(n): #rows
    for j in range(n): #columns
        if matrix[i][j] != matrix[j][i]:
            symmetric=False
            break
    if not symmetric:
        break

if symmetric:
    print("symmetric")
else:
    print("not symmetric")
```

```python
#108. Identity Matrix Check

'''
An identity matrix is a square matrix in which:

- All diagonal elements are 1.
- All non-diagonal elements are 0.
'''

n = int(input("Enter matrix size: "))

matrix = []

for i in range(n):
    row = list(map(int, input("Enter row elements: ").split()))
    matrix.append(row)

identity = True

for i in range(n): #rows
    for j in range(n): #columns
        if i == j: # i represents the row index 0, and j represents the column index 0.
            if matrix[i][j] != 1:
                identity = False
                break
        else:
            if matrix[i][j] != 0:
                identity = False
                break

    if not identity:
        break

if identity:
    print("Identity matrix")
else:
    print("Not an identity matrix")
```

```python
# 109. Row-wise and column-wise sum

r = int(input("Enter number of rows: "))
c = int(input("Enter number of columns: "))

matrix = []

for i in range(r):
    row = list(map(int, input("Enter row elements: ").split()))
    matrix.append(row)

# Row-wise sum
for i in range(r):
    total = 0

    for j in range(c):
        total += matrix[i][j]

    print("Row", i + 1, "sum:", total)

# Column-wise sum
for j in range(c):
    total = 0

    for i in range(r):
        total += matrix[i][j]

    print("Column", j + 1, "sum:", total)
```

```python
# 110. Maximum and minimum element in a matrix

r = int(input("Enter number of rows: "))
c = int(input("Enter number of columns: "))

matrix = []

for i in range(r):
    row = list(map(int, input("Enter row elements: ").split()))
    matrix.append(row)

maximum = matrix[0][0]
minimum = matrix[0][0]

for i in range(r):
    for j in range(c):
        if matrix[i][j] > maximum:
            maximum = matrix[i][j]

        if matrix[i][j] < minimum:
            minimum = matrix[i][j]

print("Maximum element:", maximum)
print("Minimum element:", minimum)
```

```python
# 111. Selection sort

arr = [64, 25, 12, 22, 11]

for i in range(len(arr)):
    min_index = i

    for j in range(i + 1, len(arr)):
        if arr[j] < arr[min_index]:
            min_index = j

    arr[i], arr[min_index] = arr[min_index], arr[i]

print("Sorted array:", arr)
```

```python
# 112. Insertion sort

arr = [12, 11, 13, 5, 6]

for i in range(1, len(arr)):
    key = arr[i]
    j = i - 1

    while j >= 0 and arr[j] > key:
        arr[j + 1] = arr[j]
        j -= 1

    arr[j + 1] = key

print("Sorted array:", arr)
```

```python
# 113. Check if an array is sorted

arr = [10, 20, 30, 40, 50]

is_sorted = True

for i in range(len(arr) - 1):
    if arr[i] > arr[i + 1]:
        is_sorted = False
        break

if is_sorted:
    print("Array is sorted")
else:
    print("Array is not sorted")
```

```python
# 114. Count even and odd numbers in an array

arr = [10, 15, 22, 33, 40, 7]

even = 0
odd = 0

for num in arr:
    if num % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even numbers:", even)
print("Odd numbers:", odd)
```

```python
# 115. Count positive, negative and zero elements

arr = [10, -5, 0, 20, -8, 0, 7]

positive = 0
negative = 0
zero = 0

for num in arr:
    if num > 0:
        positive = positive + 1
    elif num < 0:
        negative = negative + 1
    else:
        zero = zero + 1

print("Positive:", positive)
print("Negative:", negative)
print("Zeros:", zero)
```

```python
# 116. Union and intersection of two arrays

arr1 = [1, 2, 3, 4, 5]
arr2 = [4, 5, 6, 7, 8]

a=set(arr1)
b=set(arr2)

union = list( a | b )
intersection = list( a & b )

print("Union:", sorted(union))
print("Intersection:", sorted(intersection))
```

```python
# 119. Hollow square pattern

n = 5

for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# 120. Hollow right triangle pattern

n = 5

for i in range(1, n + 1):
    for j in range(1, i + 1):
        if j == 1 or j == i or i == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# 128. Write a Python program to create a simple calculator using arithmetic operators.

num1 = float(input("Enter first number: "))
operator = input("Enter operator (+, -, *, /, %, **): ")
num2 = float(input("Enter second number: "))

if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Cannot divide by zero"
elif operator == "%":
    if num2 != 0:
        result = num1 % num2
    else:
        result = "Cannot calculate modulo by zero"
elif operator == "**":
    result = num1 ** num2
else:
    result = "Invalid operator"

print("Result:", result)
```
