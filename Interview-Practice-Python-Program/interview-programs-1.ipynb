```python
#Prime Numbers without Square root

n = 19

if n < 2:
    print(n, "is not a prime number")
else:
    for i in range(2, int(n ** 0.5) + 1):
        print(i, end=" ") # end=" " prints each number on the single line, separated by spaces.
        if n % i == 0:
            print("\n", n, "is not a prime number") # \n moves the output to the next line.
            break
    else:
        print("\n", n, "is a prime number")
```

```python
#Prime Numbers using Square root

import math

n = 19

if n < 2:
    print(n, "is not a prime number")
else:
    for i in range(2, int(math.sqrt(n)) + 1):
        print(i, end=" ") # end=" " prints each number on the single line, separated by spaces.

        if n % i == 0:
            print("\n", n, "is not a prime number") # \n moves the output to the next line.
            break
    else:
        print("\n", n, "is a prime number")
```

```python
#Armstrong Number(123 - 1^3 + 2^3 + 3^3 , 12345 - 1^5 + 2^5 + 3^5 + 4^5 +5^5)

n = 153

digits_cube = len(str(n))
total = 0
temp = n

while temp > 0:
    digit = temp % 10
    total =total + digit ** digits_cube
    temp =temp // 10

if total == n:
    print("Armstrong")
else:
    print("Not Armstrong")
```

```python
s1 = "listen"
s2 = "silent"

if len(s1) != len(s2):
    print("Not Anagram")
else:
    freq = {}

    for ch in s1:
        freq[ch] = freq.get(ch, 0) + 1

    for ch in s2:
        freq[ch] = freq.get(ch, 0) - 1

    if all(count == 0 for count in freq.values()):
        print("Anagram")
    else:
        print("Not Anagram")
```

```python
#sorted the list without sorted() func

a = [10, 20, 5, 20, 15]

for i in range(len(a)):
    for j in range(i+1,len(a)):
        if a[i]>a[j]:
            a[i],a[j]=a[j],a[i]
print("sorted list:",a)
```
