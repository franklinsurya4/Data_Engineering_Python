```python
#22. GCD (HCF) of two numbers ( without function)

''' Example: Find the GCD of 12 and 18.

- Factors of 12: 1, 2, 3, 4, 6, 12
- Factors of 18: 1, 2, 3, 6, 9, 18

- Common factors: 1, 2, 3, 6

- Greatest Common Divisor = 6  '''


# 22. GCD (HCF) of two numbers

num1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))

a = num1
b = num2

while b != 0:
    remainder = a % b
    a = b
    b = remainder

print("GCD of", num1, "and", num2, "is", a)
```

```python
#23. lcm of 2 numbers (without function)

''' The LCM of 4 and 8 is 8.

Explanation:

- Multiples of 4: 4, 8, 12, 16, 20, ...
- Multiples of 8: 8, 16, 24, 32, ...

- Smallest common multiple = 8 '''

num1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))

a = num1
b = num2

while b != 0:
    remainder = a % b
    a = b
    b = remainder

gcd = a


lcm = (num1 * num2) // gcd

print("LCM of", num1, "and", num2, "is", lcm)
```

```python
#23. lcm of 2 numbers (with function)

''' The LCM of 4 and 8 is 8.

Explanation:

- Multiples of 4: 4, 8, 12, 16, 20, ...
- Multiples of 8: 8, 16, 24, 32, ...

- Smallest common multiple = 8 '''

def lcm(num1,num2):

    a = num1
    b = num2

    while b != 0:
        remainder = a % b
        a = b
        b = remainder

    gcd = a

    return (num1 * num2) // gcd

num1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))

print("LCM of", num1, "and", num2, "is", lcm(num1,num2))
```

```python
#24. Automorphic Number (without function)

''' Examples:

- 5 - 5^2 = 25 → ends with 5, so 5 is an Automorphic number.

- 6 - 6^2 = 36 → ends with 6, so 6 is an Automorphic number.

- 25 - 25^2 = 625 → ends with 25, so 25 is an Automorphic number. '''

num = int(input("Enter a number: "))

square = num * num
digits = len(str(num))

if num >= 0 and square % (10 ** digits) == num:
    print(num, "is an Automorphic number")
else:
    print(num, "is not an Automorphic number")
```

```python
# 25. Harshad (Niven) Number (without function)

''' Example: 18 is a Harshad Number.

Sum of digits: 1+8=9

Divide the number by the sum of its digits:

18/9=2

Since the remainder is 0, 18 is a Harshad Number.

Other examples: 1, 2, 9, 10, 12, 18, 20, 21, 24. '''

num = int(input("Enter a number: "))

temp = num
sum_add = 0

while temp > 0:
    digit = temp % 10
    sum_add = sum_add + digit
    temp = temp // 10

if num > 0 and num % sum_add == 0:
    print(num, "is a Harshad/Niven Number")
else:
    print(num, "is not a Harshad/Niven Number")
```

```python
#37. Spy Number in Python (Without Function)

'''
Example: 1124

- Sum of digits = 1+1+2+4=8

- Product of digits = 1 x 1 x 2 x 4=8

Since the sum and product are equal, 1124 is a Spy number.
'''

num = 1124

original_num = num

sum_digits = 0
product = 1

if num == 0:
    product = 0

while num > 0:
    digit = num % 10
    sum_digits = sum_digits + digit
    product = product * digit
    num = num // 10

if sum_digits == product:
    print(original_num, "is a Spy number")
else:
    print(original_num, "is not a Spy number")
```

```python
# 44. Power of Two (WITHOUT FUNCTION)

'''
Examples: 1, 2, 4, 8, 16, 32, 64, 128.

For example, 8 = 2^3 = 2 x2 x 2.

Suppose the input is 16.

1. 16 % 2 == 0 → divide by 2 → 8

2. 8 % 2 == 0 → divide by 2 → 4

3. 4 % 2 == 0 → divide by 2 → 2

4. 2 % 2 == 0 → divide by 2 → 1

5. Now num == 1, so it is a power of two.
'''

num = 16

if num < 1:
    print(num, "is not a power of two")
else:
    while num % 2 == 0:
        num = num // 2

    if num == 1:
        print("It is a power of two")
    else:
        print("It is not a power of two")
```

```python
#46. Happy Number (WITHOUT FUNCTION)

'''
A Happy number is a number where repeatedly replacing the number with the sum of the squares of its digits eventually produces 1.

Example: 19

1^2 +9 ^2 = 1+81 = 82

8^2 + 2^2 = 64+4 = 68

6^2 + 8^2 = 36+64 = 100

1^2 + 0^2 + 0^2 = 1

Therefore, 19 is a Happy number. '''

num = 19

n = num
seen = set()

while n != 1 and n not in seen:
    seen.add(n)
    sum_square = 0

    while n > 0:
        digit = n % 10
        sum_square = sum_square + digit ** 2
        n = n // 10

    n = sum_square

if n == 1:
    print(num, "is a Happy number")
else:
    print(num, "is not a Happy number")
```

```python
#47. Peterson Number (WITHOUT FUNCTION)

'''
A Peterson number is a number whose sum of the factorials of its digits equals the original number.
Example: 145

1! + 4! + 5! = 1 + 24 + 120 = 145

Therefore, 145 is a Peterson number.

Remember that 0! = 1. '''

num = 145

temp = num
sum_fact = 0

while temp > 0:
    digit = temp % 10

    fact = 1
    for i in range(1, digit + 1):
        fact = fact * i

    sum_fact = sum_fact + fact
    temp = temp // 10

if sum_fact == num:
    print(num, "is a Peterson number")
else:
    print(num, "is not a Peterson number")
```
