```python
#20. Perfect number (Without Function),

''' Example 1: 6 is a Perfect Number

Factors of 6, excluding 6 itself:

- Factors: 1, 2, 3

- Sum: 1 + 2 + 3 = 6

Therefore, 6 is a Perfect Number.

Example 2: 28 is a Perfect Number

- Factors: 1, 2, 4, 7, 14, excluding 28 itself:

- Sum: 1 + 2 + 4 + 7 + 14 = 28

Therefore, 28 is a Perfect Number. '''

num=int(input("enter a number"))

total=0

for i in range(1,num):
    if num%i==0:
        total=total+i

if num>0 and total==num:
    print(num,"is a perfect number")
else:
   print(num, "is not a perfect number")
```

```python
#20. Perfect Number (With Function)

''' Example 1: 6 is a Perfect Number

Factors of 6, excluding 6 itself:

- Factors: 1, 2, 3

- Sum: 1 + 2 + 3 = 6

Therefore, 6 is a Perfect Number.

Example 2: 28 is a Perfect Number

- Factors: 1, 2, 4, 7, 14, excluding 28 itself:

- Sum: 1 + 2 + 4 + 7 + 14 = 28

Therefore, 28 is a Perfect Number. '''

def perfect_num(num):
    total = 0

    for i in range(1, num):
        if num % i == 0:
            total = total + i

    if total == num:
       print(num, "is a Perfect Number")
    else:
        print(num, "is not a Perfect Number")

num = int(input("Enter a number: "))
perfect_num(num)
```

```python
#22. GCD (HCF) of two numbers ( without function)

''' Example: Find the GCD of 12 and 18.

- Factors of 12: 1, 2, 3, 4, 6, 12
- Factors of 18: 1, 2, 3, 6, 9, 18

- Common factors: 1, 2, 3, 6

- Greatest Common Divisor = 6  '''


# 22. GCD (HCF) of two numbers

num1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))

a = num1
b = num2

while b != 0:
    remainder = a % b
    a = b
    b = remainder

print("GCD of", num1, "and", num2, "is", a)
```

```python
#23. lcm of 2 numbers (without function)

''' The LCM of 4 and 8 is 8.

Explanation:

- Multiples of 4: 4, 8, 12, 16, 20, ...
- Multiples of 8: 8, 16, 24, 32, ...

- Smallest common multiple = 8 '''

num1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))

a = num1
b = num2

while b != 0:
    remainder = a % b
    a = b
    b = remainder

gcd = a


lcm = (num1 * num2) // gcd

print("LCM of", num1, "and", num2, "is", lcm)
```

```python
#23. lcm of 2 numbers (with function)

''' The LCM of 4 and 8 is 8.

Explanation:

- Multiples of 4: 4, 8, 12, 16, 20, ...
- Multiples of 8: 8, 16, 24, 32, ...

- Smallest common multiple = 8 '''

def lcm(num1,num2):

    a = num1
    b = num2

    while b != 0:
        remainder = a % b
        a = b
        b = remainder

    gcd = a

    return (num1 * num2) // gcd

num1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))

print("LCM of", num1, "and", num2, "is", lcm(num1,num2))
```

```python
#24. Automorphic Number (without function)

''' Examples:

- 5 - 5^2 = 25 → ends with 5, so 5 is an Automorphic number.

- 6 - 6^2 = 36 → ends with 6, so 6 is an Automorphic number.

- 25 - 25^2 = 625 → ends with 25, so 25 is an Automorphic number. '''

num = int(input("Enter a number: "))

square = num * num
digits = len(str(num))

if num >= 0 and square % (10 ** digits) == num:
    print(num, "is an Automorphic number")
else:
    print(num, "is not an Automorphic number")
```

```python
# 24. Automorphic number (with function)

''' Examples:

- 5 - 5^2 = 25 → ends with 5, so 5 is an Automorphic number.

- 6 - 6^2 = 36 → ends with 6, so 6 is an Automorphic number.

- 25 - 25^2 = 625 → ends with 25, so 25 is an Automorphic number. '''

def automorphic(num):

    square = num * num
    digits = len(str(num))

    if num >= 0 and square % (10 ** digits) == num:
        print(num, "is an Automorphic number")
    else:
        print(num, "is not an Automorphic number")


num = int(input("Enter a number: "))

automorphic(num)
```

```python
# 25. Harshad (Niven) Number (without function)

''' Example: 18 is a Harshad Number.

Sum of digits: 1+8=9

Divide the number by the sum of its digits:

18/9=2

Since the remainder is 0, 18 is a Harshad Number.

Other examples: 1, 2, 9, 10, 12, 18, 20, 21, 24. '''

num = int(input("Enter a number: "))

temp = num
sum_add = 0

while temp > 0:
    digit = temp % 10
    sum_add = sum_add + digit
    temp = temp // 10

if num > 0 and num % sum_add == 0:
    print(num, "is a Harshad/Niven Number")
else:
    print(num, "is not a Harshad/Niven Number")
```

```python
# 25. Harshad (Niven) Number (with function)

''' Example: 18 is a Harshad Number.

Sum of digits: 1+8=9

Divide the number by the sum of its digits:

18/9=2

Since the remainder is 0, 18 is a Harshad Number.

Other examples: 1, 2, 9, 10, 12, 18, 20, 21, 24. '''


def Harshad(num):

    temp = num
    sum_add = 0

    while temp > 0:
        digit = temp % 10
        sum_add = sum_add + digit
        temp = temp // 10

    if num > 0 and num % sum_add == 0:
        print(num, "is a Harshad/Niven Number")
    else:
        print(num, "is not a Harshad/Niven Number")

num = int(input("Enter a number: "))

Harshad(num)
```

```python
#26. Neon number  (without function)

''' Example: 9 is a Neon Number.

Step 1: Find the square of 9.

9^2 = 81

Step 2: Add the digits of 81.

8+1=9

Step 3: Compare the sum with the original number.

Since 9=9, 9 is a Neon Number. '''

num = int(input("Enter a number: "))

square = num * num
temp = square
total = 0

while temp > 0:
    digit = temp % 10
    total = total + digit
    temp = temp // 10

if num >= 0 and total == num:
    print(num, "is a Neon Number")
else:
    print(num, "is not a Neon Number")
```

```python
# 27. Perfect Square - Without Function

'''Example: 25 is a Perfect Square

5 = 25

5^2 - 5*5= 25, 25 is a Perfect Square.

Other examples: 0, 1, 4, 9, 16, 25, 36, 49, 64, 81, 100. '''

num = int(input("Enter a number: "))

if num >= 0:
    root = int(num ** 0.5)

    if root * root == num:
        print(num, "is a Perfect Square")
    else:
        print(num, "is not a Perfect Square")
else:
    print(num, "is not a Perfect Square")
```

```python
# 27. Perfect Square (with Function)

'''Example: 25 is a Perfect Square

5 = 25

5^2 = 25, 25 is a Perfect Square.

Other examples: 0, 1, 4, 9, 16, 25, 36, 49, 64, 81, 100. '''

def perfect_square(num):
    if num < 0:
        print(num, "is not a Perfect Square")
        return

    root = int(num ** 0.5)

    if root * root == num:
        print(num, "is a Perfect Square")
    else:
        print(num, "is not a Perfect Square")


num = int(input("Enter a number: "))
perfect_square(num)
```
