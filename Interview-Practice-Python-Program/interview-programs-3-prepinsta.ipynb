```python
# 58. Binary search (without function)

a = [10, 20, 30, 40, 50, 60, 70]

target = int(input("Enter the element to search: "))

low = 0
high = len(a) - 1
found = False

while low <= high:
    mid = (low + high) // 2

    if a[mid] == target:
        print("Element found at index", mid)
        found = True
        break

    elif a[mid] < target:
        low = mid + 1

    else:
        high = mid - 1

if not found:
    print("Element not found")
```

```python
# 58. Binary search (with function)

def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid

        elif arr[mid] < target:
            low = mid + 1

        else:
            high = mid - 1

    return -1


a = [10, 20, 30, 40, 50, 60, 70]

target = int(input("Enter the element to search: "))

result = binary_search(a, target)

if result != -1:
    print("Element found at index", result)
else:
    print("Element not found")
```

```python
#95. Count Set Bits in a Number

'''
Concept: A set bit is a bit whose value is 1 in the binary representation of a number.

Example:

- Number: 13
- Binary: 1101
- Number of set bits: 3
'''

num=13

count=0

while num > 0 :

    if num % 2 == 1:
        count=count+1
    num=num//2

print("Number of set bits:", count)
```

```python
#96. Reverse Words in a Sentence

'''
Example:

- Input: "Python is very easy"
- Output: "easy very is Python"
'''

text="Python is very easy"

words=text.split()

rev=[]

for i in range(len(words)-1,-1,-1):
    rev.append(words[i])

print(" ".join(rev))
```

```python
# 96. Reverse words in a sentence

sentence = "Python is very easy"

words = sentence.split()

reversed_words = words[::-1]

result = " ".join(reversed_words)

print(result)
```

```python
#97. Longest Word in a Sentence

'''
Example:

- Input: "Python programming is interesting"
- Output: programming
'''

text="Python programming is interesting"

words=text.split()

longest=""

for word in words:
    if len(word) > len(longest):
        longest=word

print(longest)
```

```python
#98. Capitalize first letter of each word

text="python programming is interesting"

words=text.split()

result=[]

for word in words:
    capitalize=word[0].upper()+word[1:]
    result.append(capitalize)

print(" ".join(result))
```

```python
#99. Count uppercase, lowercase, digits and special characters

text = "Hello@123"

upp_count=0
low_count=0
dig_count=0
sc_count=0

for ch in text:
    if ch.isupper():
        upp_count=upp_count+1
    elif ch.islower():
        low_count=low_count+1
    elif ch.isdigit():
        dig_count=dig_count+1
    else:
        sc_count=sc_count+1

print("uppercase count:",upp_count)
print("lowercase count:",low_count)
print("digits count:",dig_count)
print("special characters count:",sc_count)
```

```python
#100. Remove vowels from a string

text="Franklin Surya"

remove_vowels=""

for ch in text:
    if ch.lower() not in "aeiou":
        remove_vowels=remove_vowels+ch

print(remove_vowels)
```
