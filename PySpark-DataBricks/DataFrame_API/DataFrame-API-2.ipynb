## I/O in Apache Spark (Input and Output):

I/O stands for `Input/Output`. It is a process of `reading data from a source (Input)`, `processing it usinf spark`, and `writing the results to a destination (Output) such as csv, json, database, cloud`.
- Input (I): reading data from a source.
- Output (O): writing the results to a destination.

### Real-Time Example

- **Input:** `sales.csv` contains daily order details.
- **Read:** Spark loads the CSV into a DataFrame.
- **Clean:** Remove records with missing or invalid values.
- **Transform:** Calculate total sales and summarize sales by product.
- **Output:** Save the cleaned data in Parquet format to `sales_processed`.
- **Business use:** Use the processed data to build sales reports and dashboards.


## Step 1: Read the CSV file:

````python
df = spark.read.csv(
    "/Workspace/Data-Enginnering/Data-Engineering-Python/PySpark-DataBricks/Source/sales.csv",
    header=True,
    inferSchema=True
)

df.show()
````

Explanation:
- `spark.read` — starts the read operation.
- `csv()` — reads a CSV file.
- `header=True` — treats the first row as column names.
- `inferSchema=True` — asks Spark to infer column data types.

## Step 2: Process the data:

Remove records where Amount is null and calculate a new column for 18% tax

````python
from pyspark.sql.functions import col

clean_df = df.dropna(subset=["Amount"])

result_df = clean_df.withColumn(
    "Tax",
    col("Amount") * 0.18
)

result_df.show()
````

## Step 3: Output — Write the processed data:

````python
result_df.write \
    .mode("overwrite") \
    .parquet(
        "/Volumes/workspace/default/data_volume/sales_processed"
    )
````

Explanation:
- `write` — starts the output operation.
- `mode("overwrite")` — replaces existing output data at the destination.
- `parquet()` — writes the DataFrame in Parquet format.

## Common Spark I/O Formats

| Format | Read Example | Write Example |
|---|---|---|
| CSV | `spark.read.csv(path)` | `df.write.csv(path)` |
| JSON | `spark.read.json(path)` | `df.write.json(path)` |
| Parquet | `spark.read.parquet(path)` | `df.write.parquet(path)` |
| JDBC Database | `spark.read.jdbc(url, table, properties)` | `df.write.jdbc(url, table, properties)` |
| Delta Table | `spark.read.format("delta").load(path)` | `df.write.format("delta").save(path)` |


## Important interview questions:


### # Q1. What is I/O in Spark?

I/O is the process of reading data from input sources and writing processed results to output destinations.

### # Q2. What is the difference between spark.read and df.write?

- spark.read reads data into Spark.
- df.write saves a DataFrame to a destination.

### # Q3. Why do we use Parquet instead of CSV?

Parquet is a columnar storage format that supports efficient column reads, compression, and schema information. It is often a better choice for analytical data pipelines.

### # Q4. Does Spark immediately execute transformations after reading a file?

Not necessarily. Spark generally builds a lazy execution plan for transformations and performs the required work when an action or output operation triggers execution.

### # Remember: 

Input → Read → Transform → Write → Output. This is the basic I/O workflow in a Spark data engineering pipeline.
