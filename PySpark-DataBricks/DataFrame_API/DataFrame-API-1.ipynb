## DataFrame API:

DataFrame API is a `Spark interface` which is used to `process structured large data using rows, columns, and built-in functions`. It supports operations like `filtering, selecting, grouping, joining, and aggregating data`.

## Step 1: Create a DataFrame:

````python
data = [
    (101, "Arun", "IT", 45000),
    (102, "Priya", "HR", 28000),
    (103, "Kumar", "IT", 55000),
    (104, "Divya", "HR", 35000)
]

columns = ["EmployeeID", "Name", "Department", "Salary"]

df = spark.createDataFrame(data, columns)

df.show()
````

`spark.createDataFrame()` is used to creates a DataFrame from the supplied data and column names.

## Step 2: Select specific columns:

````python
df.select("Name", "Salary").show()
````

`select()` is used to retrieves the required columns.

`Real-time use:` Display employee names and salaries without showing unnecessary information

## Step 3: Filter employees by salary:

````python
df.filter(df.Salary > 30000).show()
````

`filter()` used to filter rows and columns based on condition.

## Step 4: Add a new column:

````python
from pyspark.sql.functions import col

df.withColumn(
    "AnnualSalary",
    col("Salary") * 12
).show()
````

`withColumn()` is used to adds a new column or replaces an existing column.

## Step 5: Group and calculate total salary:

````python
from pyspark.sql.functions import sum

df.groupBy("Department").agg(
    sum("Salary").alias("TotalSalary")
).show()
````

## Important DataFrame API Operations

| API Operation | Simple Definition | Real-Time Example |
|---|---|---|
| `select()` | Choose specific columns | Display employee name and salary |
| `filter()` / `where()` | Filter rows based on a condition | Employees with salary above ₹30,000 |
| `withColumn()` | Add or replace a column | Calculate annual salary |
| `drop()` | Remove columns | Remove unnecessary employee data |
| `dropna()` | Remove rows with null values | Remove records missing required data |
| `groupBy()` | Group rows based on a column | Group employees by department |
| `agg()` | Calculate summary values | Calculate total salary per department |
| `join()` | Combine two DataFrames using a related column | Combine employee and department data |
| `orderBy()` | Sort rows | Display highest salary first |
| `distinct()` | Return distinct rows | Remove duplicate records |
| `count()` | Count the number of rows | Count total employees |
| `show()` | Display DataFrame rows | Inspect processed employee data |

## DataFrame API vs SQL API

| DataFrame API | Spark SQL |
|---|---|
| Uses Python or Scala methods | Uses SQL queries |
| `df.filter(col("Salary") > 30000)` | `SELECT * FROM employees WHERE Salary > 30000` |
| Uses DataFrame operations such as `select()`, `filter()`, and `groupBy()` | Uses SQL commands such as `SELECT`, `WHERE`, and `GROUP BY` |
| Useful for programmatic data pipelines | Useful for SQL-based data analysis |
| Supports dynamic transformations using code | Supports querying registered tables and views |
| Example: `df.select("Name", "Salary")` | Example: `SELECT Name, Salary FROM employees` |

## Q: Why do we use the DataFrame API in PySpark?

DataFrame API is used to `process structured large datasets`. It provides operations for `filtering, selecting, joining, aggregating, and transforming data`. Spark can optimize the `execution plan and distribute processing` across a cluster.
