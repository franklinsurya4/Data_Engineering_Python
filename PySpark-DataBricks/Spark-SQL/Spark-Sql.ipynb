
## Spark SQL:

## 1. Spark SQL:

Spark SQL is a `Spark module` used to `process and analyze structured and semi-structured data` using `SQL queries and DataFrame operations`.


## Why do we use Spark SQL?

- Spark SQL is used to `process and analyze structured data` using `SQL queries and DataFrame operations`. 
- It is `faster` because Spark optimizes queries using the `Catalyst Optimizer`.


**Real-time example:**

**Suppose an e-commerce company has 10 million order records and wants to calculate revenue for each product.**

```python
%sql
CREATE TABLE orders (
    order_id INT,
    product_id INT,
    amount DOUBLE
);

INSERT INTO orders VALUES
(1, 101, 5000),
(2, 102, 3000),
(3, 101, 2000),
(4, 103, 4000),
(5, 102, 1500),
(6, 103, 2500);
```

```python
%sql
SELECT * FROM orders;
```

```python
%sql
SELECT
    product_id,
    SUM(amount) AS total_revenue
FROM orders
GROUP BY product_id;
```


## 2. Temporary Views in Spark SQL:

- A Temporary View is a `temporary table` created from ` DataFrame operations or SQL query`. 
- It is used to `access and process data` using `SQL queries` and `without storing data permanently`.


## Types of Temporary Views:

- Local Temporary View
- Global Temporary View


### A. local Temporary View: :

- A **local temporary view** means `dataframes can be available only within the Spark session`.
- Access by `view name`.
- Lifetime is	`Until Spark session ends`.	
- Storage is	`Temporary`.
- Created by	`createOrReplaceTempView()`.

### B. Global Temporary View:

- A **global temporary view** means `dataframes can be available across Spark sessions within the same Spark application`.
- Access	by `global_temp.view_name`
- Lifetime is `Until Spark application ends`.
- Storage is	`Temporary`.
- Created by `createOrReplaceGlobalTempView()`.


### A.local temporary view :


**Real Time Example:**


**Step 1: Create employee DataFrame:**

```python
data = [
    (1, "Franklin", "IT", 30000),
    (2, "Arun", "HR", 25000),
    (3, "Kumar", "IT", 40000),
    (4, "Priya", "Sales", 35000)
]

columns = ["id", "name", "department", "salary"]

df = spark.createDataFrame(data, columns)
```


**Step 2: Create Local Temporary View:**

```python
df.createOrReplaceTempView("employees_view")
```


**Step 3: Query using Spark SQL:**

```python
%sql
SELECT * FROM employees_view;
```


**Step 4: Find department-wise salary:**

```python
%sql
SELECT
    department,
    SUM(salary) AS total_salary
FROM employees_view
GROUP BY department;
```


## B. Global Temporary View:






**Real Time Example:**


**Create a global temporary view:**

```python
df.createOrReplaceGlobalTempView("employees_global")
```


**Access it using SQL:**

```python
%sql
SELECT * FROM global_temp.employees_global;
```


### Local vs Global Temporary View

| Feature | Local Temporary View | Global Temporary View |
|---|---|---|
| Scope | Current Spark session | Same Spark application |
| Access | Direct view name | `global_temp.view_name` |
| Lifetime | Until Spark session ends | Until Spark application ends |
| Storage | Temporary | Temporary |
| Permanent Table? | No | No |
| Visibility | Only within the current session | Accessible across sessions in the same application |
| Creation | `createOrReplaceTempView()` | `createOrReplaceGlobalTempView()` |

## 3. Regular Views vs Materialized Views:

### A. Regular View:

- A **Regular View** is a `virtual table` created using a `SQL query`. 
- It is used to stores the `query definition`.

### B. Materialized View:
- A **materialized view** is `different` from a `temporary or regular view`.
- It is used to stores the `query results`.


### Real-Time Example for Regular and Materialized view:

**An e-commerce company wants to calculate total sales for each product.**


### Step 1: Create sample data:

```python
data = [
    (1, 101, 5000),
    (2, 102, 3000),
    (3, 101, 2000),
    (4, 103, 4000),
    (5, 102, 1500)
]

columns = ["order_id", "product_id", "amount"]

orders_df = spark.createDataFrame(data, columns)

orders_df.createOrReplaceTempView("orders")
```


### Step 2: Create a regular view using SQL:

```python
%sql
CREATE OR REPLACE VIEW product_sales_view AS
SELECT
    product_id,
    SUM(amount) AS total_revenue
FROM orders
GROUP BY product_id;
```


### Step 3: Query the view:

```python
%sql
SELECT * FROM product_sales_view;
```


### Create a MATERIALIZED view using SQL:

```python
%sql
CREATE OR REPLACE MATERIALIZED VIEW product_sales_mv AS
SELECT
    product_id,
    SUM(amount) AS total_revenue
FROM orders
GROUP BY product_id;
```


### Query the view:

```python
%sql
SELECT * FROM product_sales_mv;
```

**B. Materialized View:**

- A **materialized view** is `different` from a `regular Spark temporary or regular view`.
- It is used to stores the `query results`.

**Q2. Which is faster regular view or materialized view?**

- A `materialized view` is faster.
- Because It is `different` from a `regular Spark temporary or regular view`.
- It is used to stores the `query results`.

**Q3. Does a regular view store data?**

- No. 
- It stores the `query definition`, not the `result set`.

**Q4. Does a materialized view always contain the latest data?**

- `No`, **materialized view** does not always contain the latest data. 
- Its `latest data` depends on `when and how` the results are `refreshed`.

**Q5. When should we use a materialized view?**

 We should use a **materialized view** in `expensive aggregations or analytical queries` and to reduce `repeated computation`.


## 4. Managed Tables vs External Tables:


### A. Managed Table: 

- A **managed table** is a table managed by `databricks` for both ` metadata` and `underlying data`. 
- When we drop the table,` both metadata and underlying data are deleted`.

### B. External Table: 

- An **external table** is a table managed by`databricks` for both ` metadata` and `underlying data`, but the `underlying data is stored at an external storage location`. 
- When we drop the table, `only metadata is deleted`; the underlying data remains.


### Real-Time Example:


### Managed Table: 

**An e-commerce company stores its internal sales data in a managed table. When the table is dropped, its underlying data is also deleted.**

### External Table: 

**A company stores customer transaction files in cloud storage. Databricks creates an external table pointing to that location. If the table is dropped, the original transaction files remain in cloud storage.**


### PySpark Example – Managed Table:

```python
data = [
    (1, "Laptop", 55000),
    (2, "Mobile", 15000),
    (3, "Mouse", 500)
]

columns = ["product_id", "product", "amount"]

df = spark.createDataFrame(data, columns)

df.write.mode("overwrite").saveAsTable("managed_products")
```

```python
%sql
SELECT * FROM managed_products;
```

```python
%sql
DROP TABLE managed_products;
```


### Result: 
**The managed table and its underlying data are deleted.**


### PySpark Example – External Table

First, write data to a storage location:

```python
df.write.format("delta") \
    .mode("overwrite") \
    .save("/Volumes/workspace/default/data_volume/external_products")
```

### Create an external table:

```python
%sql
CREATE TABLE external_products
USING DELTA
LOCATION '/Volumes/workspace/default/data_volume/external_products';
```


### Query:

```python
%sql
SELECT * FROM external_products;
```


### Drop:

```python
%sql
DROP TABLE external_products;
```


### Example 4: CASE WHEN


**DSL:**

```python
from pyspark.sql.functions import when

df.withColumn(
    "salary_category",
    when(col("salary") > 40000, "High")
    .otherwise("Low")
).show()
```


**SQL:**

```python
%sql
SELECT *,
    CASE
        WHEN salary > 40000 THEN 'High'
        ELSE 'Low'
    END AS salary_category
FROM employees;
```


### Example 5: Window Functions



**DSL:**

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number

window_spec = Window.partitionBy("department") \
                    .orderBy(col("salary").desc())

df.withColumn(
    "rank",
    row_number().over(window_spec)
).show()
```


**SQL:**

```python
%sql
SELECT *,
    ROW_NUMBER() OVER (
        PARTITION BY department
        ORDER BY salary DESC
    ) AS rank
FROM employees;
```


## Important Concepts for Interview:


**What is MERGE?**

- MERGE is used to `insert, update or delete records` in a `target table` based on `matching conditions`. 
- It is commonly used in `incremental ETL` and `upsert operations`.


### Step 1: Create the target table:

```python
%sql
CREATE OR REPLACE TABLE target_table (
    id INT,
    name STRING,
    salary INT
);

INSERT INTO target_table VALUES
(1, 'Franklin', 40000),
(2, 'Surya', 55000),
(4, 'Ravi', 35000);
```


### Step 2: Create the source table:

```python
%sql
CREATE OR REPLACE TABLE source_table (
    id INT,
    name STRING,
    salary INT
);

INSERT INTO source_table VALUES
(1, 'Franklin', 50000),
(2, 'Surya', 60000),
(3, 'Arun', 45000);
```


### Step 3: Execute MERGE INTO:

```python
%sql

MERGE INTO target_table t
USING source_table s
ON t.id = s.id

WHEN MATCHED THEN
    UPDATE SET
        t.name = s.name,
        t.salary = s.salary

WHEN NOT MATCHED THEN
    INSERT (id, name, salary)
    VALUES (s.id, s.name, s.salary);
```


### Step 4: Verify the result:

```python
%sql
SELECT * FROM target_table
ORDER BY id;
```


### What is Time Travel?

Time Travel means to `access previous versions of a Delta table`.


### Step 1: Save employees_df as a Delta table:

```python
employees_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("employees_delta")
```


### Step 2: Verify the Delta table:

```python
%sql

DESCRIBE DETAIL employees_delta;
```


### Step 3: Insert new data:

```python
%sql

INSERT INTO employees_delta
VALUES
(6, 'Vijay', 'IT', 45000),
(7, 'Divya', 'HR', 32000);
```


### Step 4: Check table history:

```python
%sql

DESCRIBE HISTORY employees_delta;
```


### Step 5: Perform Time Travel:

```python
%sql

SELECT *
FROM employees_delta VERSION AS OF 0;
```
