
Statistical built-in functions are used` to analyze numerical data, measure relationships between columns, identify frequent values, and summarize datasets`.


## A. corr():

**corr() **is used `to calculates the Pearson correlation coefficient between two numeric columns`, measuring `the strength and their linear relationship`.

```python
correlation_value = sales_df.stat.corr(
    "quantity",
    "net_revenue"
)

print("Quantity-Revenue Correlation:", correlation_value)
```


### B. cov():

**cov()** is used to` calculates the sample covariance between two numeric columns`, indicating whether t`heir values are move together or in opposite directions`.

```python
covariance_value = sales_df.stat.cov(
    "quantity",
    "net_revenue"
)

print("Quantity-Revenue Covariance:", covariance_value)
```


### C. sample():

**sample()** is used to `returns a random fraction of rows from a DataFrame` using a `specified sampling fraction`.

```python
sample_sales_df = sales_df.sample(
    withReplacement=False,
    fraction=0.5,
    seed=42
)

display(sample_sales_df)
```


### D. freqItems():

**freqItems()** is used to `identifies frequent items in one or more columns` using a `probabilistic algorithm`.

```python
frequent_products_df = sales_df.stat.freqItems(
    ["product_name", "category"],
    support=0.3
)

display(frequent_products_df)
```


**Business use: Explore frequently occurring product names or categories.**


### E. summary():

**summary()** is used to `generates summary for a DataFrame, including counts, means, standard deviations, minimums, and maximums` for applicable columns.

```python
sales_summary_df = sales_df.select(
    "quantity",
    "unit_price",
    "discount",
    "gross_revenue",
    "net_revenue"
).summary()

display(sales_summary_df)
```


**Business use: Understand sales distributions, identify unusually high values, and investigate data-quality issues.**
