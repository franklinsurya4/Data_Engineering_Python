
### GroupBy Aggregations:

- `groupBy()` is used to **group rows that have the same value in one or more columns**.

- `Aggregation` is used to `calculate values such as total, average, count, minimum, and maximum` for each group.

```python
%run ./03_Joins
```


### 1. Total Revenue by City:

```python
city_revenue_df = sales_df.groupBy( "city" ).agg(
    F.sum("net_revenue").alias("total_revenue"),
    F.countDistinct("order_id").alias("total_orders"),
    F.countDistinct("customer_id").alias("total_customers"),
    F.avg("net_revenue").alias("average_item_revenue")
)

display(city_revenue_df.orderBy(F.desc("total_revenue")))
```


**Business question: Which customer cities generate the most sales?**


### 2. Revenue by Product Category:

```python
category_revenue_df = sales_df.groupBy( "category" ).agg(
    F.sum("net_revenue").alias("total_revenue"),
    F.sum("quantity").alias("units_sold"),
    F.countDistinct("order_id").alias("orders_count"),
    F.avg("unit_price").alias("average_product_price")
)

display(category_revenue_df)
```


**Business question: Which product categories perform best?**


### 3. Monthly Sales Analysis:

```python
monthly_sales_df = sales_df.groupBy( "order_month").agg(
    F.sum("net_revenue").alias("monthly_revenue"),
    F.countDistinct("order_id").alias("monthly_orders"),
    F.sum("quantity").alias("units_sold")
)

display(monthly_sales_df.orderBy("order_month"))
```


**Business use: Track sales growth month by month.**


### 4. Filter Delivered Orders:

```python
delivered_sales_df = sales_df.filter(
    F.col("order_status") == "Delivered"
)

delivered_city_sales_df = delivered_sales_df.groupBy(
    "city"
).agg(
    F.sum("net_revenue").alias("delivered_revenue"),
    F.countDistinct("order_id").alias("delivered_orders")
)

display(delivered_city_sales_df)
```
