
## Set Operations:


- Set operations is used to `combine or compare the rows of two DataFrames`.
- Common operations include `union(), unionByName(), intersect(), subtract(), and exceptAll()`.


## Set Operations,Rollup, Cube are started here:


### A. union():

- **union()** is used to `combines the rows of two DataFrames`.
- It does not `remove duplicate rows automatically`.

```python
january_sales_df = sales_df.filter(
    F.month("order_date") == 1
).select(
    "order_id", "customer_id", "net_revenue"
)

february_sales_df = sales_df.filter(
    F.month("order_date") == 2
).select(
    "order_id", "customer_id", "net_revenue"
)

combined_sales_df = january_sales_df.union(
    february_sales_df
)

display(combined_sales_df)
```


**Real-time use: Combine monthly sales datasets into one dataset.**


### B. unionByName():

**unionByName(**) is used to `combines the rows of two DataFrames by matching column names`.

```python
january_reordered_df = january_sales_df.select(
    "customer_id", "order_id", "net_revenue"
)

combined_by_name_df = january_sales_df.unionByName(
    january_reordered_df.select(
        "order_id", "customer_id", "net_revenue"
    )
)

display(combined_by_name_df)
```


### C. intersect():

**intersect()** is used to` returns the distinct rows and that rows are common to both DataFrames`.

```python
january_customers_df = january_sales_df.select(
    "customer_id"
).distinct()

february_customers_df = february_sales_df.select(
    "customer_id"
).distinct()

repeat_customers_df = january_customers_df.intersect(
    february_customers_df
)

display(repeat_customers_df)
```


## E. exceptAll():

- **exceptAll()** is used to `returns rows from the first DataFrame` and `that rows are not matched by rows in the second DataFrame`.
- preserving `duplicate counts`.

```python
january_only_df = january_customers_df.exceptAll(
    february_customers_df
)

display(january_only_df)
```


## Rollup and Cube:

- **rollup() and cube()** are `aggregation methods` used to `calculate grand totals and subtotals` across multiple columns.
- They are commonly used in `reports, business intelligence, and data warehousing`.


### rollup():

- rollup() is a `aggregation method`.
- rollup() is a `hierarchical aggregation` used to `calculating subtotals and grand total` on specified columns.


**Calculate revenue by city, membership, and overall totals.**

```python
rollup_df = sales_df.rollup(
    "city",
    "membership"
).agg(
    F.sum("net_revenue").alias("total_revenue")
)

display(
    rollup_df.orderBy(
        "city",
        "membership"
    )
)
```


## cube():

- **cube()** is a `aggregation method`.
- **cube()** is a `multidimensional aggregation `used to `calculating subtotals and grand total for all combinations specified group` of columns.

```python
cube_df = sales_df.cube(
    "city",
    "membership"
).agg(
    F.sum("net_revenue").alias("total_revenue")
)

display(cube_df)
```
