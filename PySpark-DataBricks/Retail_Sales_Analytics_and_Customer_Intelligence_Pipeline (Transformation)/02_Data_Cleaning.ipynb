
### Data Cleaning and Curation process starts here:


### 1. Handle Null Values:

using **coalesce()** is used `to return a first non-null value from a list of columns or expressions`.

```python
customers_clean_df = customers_df.withColumn(
    "city",
    F.coalesce(F.col("city"), F.lit("Unknown"))
)

display(customers_clean_df)
```


**Real-time use: Replace missing customer cities with "Unknown" so those records can still be processed.**


### 2. Use trim() to Remove Extra Spaces:
using **trim()** is used `to removes leading and trailing spaces from a string`.

```python
customers_clean_df = customers_clean_df.withColumn(
    "customer_name",
    F.trim(F.col("customer_name"))
)
```


**Real-time use: Standardize customer names imported from external systems.**


### 3. Use when() and otherwise():

- **when()** is used `to applies a value when a condition is true`.
- **otherwise()** provides `a fallback when the condition is false or no condition matches`.

```python
customers_clean_df = customers_clean_df.withColumn(
    "customer_segment",
    F.when(F.col("membership") == "Gold", "Premium")
     .when(F.col("membership") == "Silver", "Regular")
     .otherwise("Basic")
)

display(customers_clean_df)
```


**Real-time use: Convert membership categories into business-friendly customer segments.**


### 4. Use lit() to Add Metadata:

**lit()** is used` to creates a literal column or expression from a constant value`.

```python
customers_clean_df = (
    customers_clean_df
    .withColumn("SourceSystem", F.lit("Retail_ERP"))
    .withColumn("CountryCode", F.lit("IN"))
    .withColumn("ProcessingStatus", F.lit("PROCESSED"))
)

display(customers_clean_df)
```


**Real-time use: Add source-system and processing metadata to support data lineage.**


### 5. Use cast() to Change Data Types:

**cast()** is used `to converts a column or expression into a specified data type`.

```python
products_clean_df = products_df.withColumn(
    "unit_price",
    F.col("unit_price").cast("decimal(12,2)")
)

display(products_clean_df)
```


**Real-time use: Standardize financial values to an appropriate decimal type.**


### 6. Use substr() and concat():

- **substr()** is used `to extracts a substring from a string using a starting position and a length`.
- **concat()** is used `to combines multiple columns or expressions into one value`.

```python
customers_clean_df = (
    customers_clean_df.withColumn(
        "customer_initial",
        F.substring(F.col("customer_name"), 1, 1)
    ).withColumn(
        "customer_label",
        F.concat(
            F.col("customer_name"),
            F.lit(" - "),
            F.col("membership")
        )
    )
)

display(customers_clean_df)
```


**Real-time use: Generate customer labels for reports.**


### 7. Use isNull() to Identify Missing Values:

```python
missing_city_df = customers_df.filter(
    F.col("city").isNull()
)

display(missing_city_df)
```


**Real-time use: Identify incomplete customer records for data-quality reporting.**

### 8. Convert Strings to Dates:

```python
orders_clean_df = orders_df.withColumn(
    "order_date",
    F.to_date("order_date", "yyyy-MM-dd")
)

display(orders_clean_df)
```


**Real-time use: Ensure order dates are stored as dates so they can be filtered, grouped by month, and used in window functions.**
