
## Spark SQL Interview Questions:


**Q1. What is Spark SQL?**

**Spark SQL** is a `Spark module` used to `process and analyze structured and semi-structured data` using `SQL queries and DataFrame operations`.

**Example:**

SELECT * FROM employees;


**Q2. Why do we use Spark SQL?**

- **Spark SQL** is used to `process and analyze structured data` using `SQL queries and DataFrame operations`. 
- It is `faster` because Spark optimizes queries using the `Catalyst Optimizer`.

**Example:** 

**A company uses Spark SQL to calculate the total sales of millions of transactions.**

SELECT SUM(amount) AS total_sales
FROM sales;


**Q3. What is the Catalyst Optimizer?**

- **Catalyst Optimizer** is `Spark SQL's query optimizer`. 
- It improves the `query execution plan` to help `queries run efficiently`.

**Example:**

SELECT name

FROM employees

WHERE salary > 50000;


**Q4. What is Tungsten?**

**Tungsten** refers to `Spark SQL's query execution engine` that help `reduce memory overhead` and `improve CPU efficiency`.

**Example:** 

**Spark can use efficient memory representations and code generation to execute supported operations more efficiently.**


**Q5. What is declarative programming?**

**Declarative programming** means specify the `what result we want` instead of `writing every step needed to produce it`.

**Example:**

SELECT department, COUNT(*)

FROM employees

GROUP BY department;

**We specify that we want the employee count for each department. Spark determines how to execute the query.**


**Q6. What is unified processing in Spark?**

**Unified processing** means `Spark to perform different data-processing tasks within one application`.

**Example:** 

**Read a CSV file, clean the data using PySpark, run SQL queries, and save the result to a Delta table.**


**Q7. What is federated querying?**

**Federated querying** means `accessing data from multiple underlying sources` through `common query interface`.

**Example:**

**A supported Spark setup might query data from a database and a file-based source within one application.**


**Q8. What is a temporary view?**

- A Temporary View is a `temporary table` created from` DataFrame operations or SQL query`.
- It is used to `access and process data` using `SQL queries` and without storing data `permanently`.

**Types of Temporary Views:**

- Local Temporary View
- Global Temporary View

**Example:**

employees_df.createOrReplaceTempView("employees")

SELECT * FROM employees;


**Q9. What is the difference between local and global temporary views?**

**local Temporary View:**

- A **local temporary view** means `dataframes can be available only within the Spark session`.
- Access by `view name`.
- Lifetime is` Until Spark session ends`.
- Storage is `Temporary`.
- Created by `createOrReplaceTempView()`.

**Example:**

employees_df.createOrReplaceTempView("employees")

SELECT * FROM employees;

**B. Global Temporary View:**

- A **global temporary view** means `dataframes can be available across Spark sessions within the same Spark application`.
- Access by `global_temp.view_name`
- Lifetime is `Until Spark application ends`.
- Storage is `Temporary`.
- Created by `createOrReplaceGlobalTempView()`.

**Example:**

employees_df.createOrReplaceGlobalTempView("employees")

SELECT * FROM global_temp.employees;


**Databricks note:** 

Global temporary views may be `unsupported on serverless compute`. Use a local temporary view for same-session work.


**Q10. Can a temporary view survive a Spark session restart?**

- No.
- A local temporary view disappears when its Spark session ends. A global temporary view disappears when its Spark application ends.

**Example: If you restart the session, you may need to register the temporary view again.**


**Q11. Does creating a temporary view duplicate the data?**

- No. 
- Creating a temporary view `does not normally copy or persist` the underlying data.

**Example:**

employees_df.createOrReplaceTempView("employees")


**Q12. What is the difference between a temporary view and a CTE?**

A **temporary view** can be reused across queries while it exists. 

**Example:**

SELECT * FROM employees;

A **CTE (Common Table Expression)** is a `temporary named result set`.
We define **CTE** using the `WITH AS` keyword and use inside a `single SQL query`.

**Example:**

WITH high_salary AS (

    SELECT *

    FROM employees

    WHERE salary > 50000

)

SELECT * FROM high_salary;

**Q23. What is partitioning?**

- **Partitioning** means dividing `large data into smaller parts`, Spark can process them in `parallel`.
- This can helps spark `to skips unnecessary data` during queries.

**Example:**

**Suppose a sales table is partitioned by year.**

CREATE TABLE sales_partitioned (
    id INT,
    amount DOUBLE,
    year INT
)

**USING DELTA:**

PARTITIONED BY (year);


**Q24. What is partition pruning?**

**Partition pruning** is an `optimization technique` in Spark that `reads only the required partitions` instead of `scanning all partitions`.

**Example:**

SELECT *
FROM sales_partitioned
WHERE year = 2026;

If the table is partitioned by year, Spark can skip partitions belonging to other years.


**Q25. What is the difference between repartition() and coalesce()?**

- repartition() is used to `increases or decreases partitions` by redistributing data across partitions. 
- coalesce() is used to return a` first non-null value` from a `list of columns or expressions`.


**Q26. What is the difference between partitioning and bucketing?**

**partitioning:**

- **Partitioning** means divides the `large data into smaller parts`, Spark can process them in `parallel`.
- This can helps spark `to skips unnecessary data` during queries.

**bucketing:**

**Bucketing** divides `table rows into a fixed number of buckets` using a bucketing column, typically through `hash-based distribution`.


**Q27. What are the disadvantages of excessive partitioning?**

**Excessive partitioning** means `creating too many small partitions` instead of `an appropriate number of partitions`.

**Disadvantages:**

- Too many tasks
- More overhead
- Small files problem
- Increased metadata
- Poor performance


**Q28. What is data skew?**

**Data skew** occurs `when data is distributed unevenly across partitions`.

**Example:**

**Suppose we have 1 million employee records distributed across 4 partitions.**

**Balanced data distribution:**

**Partition 1** - 250K records
**Partition 2** - 250K records
**Partition 3** - 250K records
**Partition 4** - 250K records

**Skewed data distribution:**

**Partition 1 **- 100K records
**Partition 2 **- 100K records
**Partition 3** - 100K records
**Partition 4** - 700K records


**Q29. What is bucketing?**

**Bucketing** is the process of divides the `table rows into a fixed number of buckets` using a bucketing column, typically through `hash-based distribution`.

**Example:**

**A table may distribute employee records into eight buckets based on employee_id.**

CLUSTERED BY (employee_id)
INTO 8 BUCKETS


**Q30. How do you create a bucketed table?**

- We can create a **bucketed table** using the `CLUSTERED BY` in SQL or `bucketBy()` in PySpark, 
- Which is used to distributes `data into a fixed number of buckets` based on the `specified column to improve query performance`, especially for `joins`.


**Q31. How can bucketing improve join performance?**

- **Bucketing** improves join performance by `distribute data into a fixed number of buckets` based on a common column, 
- Which can `reduce data shuffling` during joins.


**Q32. Is bucketing always better than partitioning?**

- No, 
- **Bucketing** is **not always better** than **Partitioning**. 
- **Partitioning** is useful for `filtering and skipping unnecessary data`,
- **Bucketing** is useful for `distributing data and optimizing joins` on commonly used columns. 
- The choice depends on the `query and data access` patterns.

**Q38. What is MERGE in Spark SQL?**

- **MERGE** is used to `insert, update or delete records` in a `target table` based on `matching conditions`. 
- It is commonly used in `incremental ETL` and `upsert operations`.

**Example:**

MERGE INTO target_table t
USING source_table s
ON t.id = s.id

WHEN MATCHED THEN
    UPDATE SET
        t.name = s.name,
        t.salary = s.salary

WHEN NOT MATCHED THEN
    INSERT (id, name, salary)
    VALUES (s.id, s.name, s.salary);


**Q39. What is the difference between INSERT, UPDATE, and MERGE?**

- **INSERT**: `Adds` new rows.
- **UPDATE**: `Modifies` existing rows.
- **MERGE**: Can `insert, update, and delete rows` based on matching conditions.


**Q40. What happens if the source table contains duplicate keys during MERGE?**

- If the source table contains **duplicate keys** during MERGE, then `MERGE operation can fail`.
- Because `multiple rows in source table are trying to update the same row in target table`. 
- Delta Lake cannot determine which `source record should be used`.

**Remove duplicate keys from the source before performing MERGE:**

from pyspark.sql.window import Window
from pyspark.sql.functions import row_number, col, desc

window_spec = Window.partitionBy("id").orderBy(desc("updated_at"))

source_df = source_df.withColumn(
    "rn", row_number().over(window_spec)
).filter(col("rn") == 1).drop("rn")


**Q41. What is time travel in Delta Lake?**

**Time Travel** means to `access previous versions of a Delta table` to `view or restore` historical data.

Using: 

- `VERSION AS OF` in SQL 
- `versionAsOf` in PySpark.

**Steps:**

1. Save employees_df ( Dataframe ) as a Delta table.
2. Verify the Delta table using ( DESCRIBE DETAIL employees_delta; ) .
3. Insert new data.
4. Check table history.
5. Perform Time Travel.

**Q42. How do you query an earlier version of a Delta table?**

Use: 

- `VERSION AS OF` in SQL 
- `versionAsOf` in PySpark.

**SQL example:**

SELECT *
FROM employees VERSION AS OF 5;

**PySpark example:**

df = (spark.read.format("delta").option("versionAsOf", 5).load("/Volumes/workspace/default/data_volume/employees"))
display(df)

**Q43. What is the difference between time travel and a backup?**

**Time Travel** means `accessing previous versions of a Delta table` to `view or restore` historical data.

**example:**

SELECT *
FROM employees VERSION AS OF 2;

**Backup** means `creating a separate copy of data` and `storing it safely for disaster recovery or long-term protection`.

**example:**

RESTORE TABLE employees
TO VERSION AS OF 2;

## Practical SQL Coding Questions with Answers:


**These six exercises cover common SQL interview patterns.**

Use the sample table employees with columns employee_id, name, department, and salary.


**Q55. How do you find the second-highest salary?**

SELECT MAX(salary) AS second_highest_salary

FROM employees

WHERE salary < (

    SELECT MAX(salary)

    FROM employees

);

**Example:**

`Salaries`: 30000, 40000, 60000

`Result`: 40000

**This returns the second-highest distinct salary, assuming at least two distinct non-null salaries exist.**


**Q56. How do you find duplicate employee IDs?**

SELECT employee_id, COUNT(*) AS total

FROM employees

GROUP BY employee_id

HAVING COUNT(*) > 1;


**Example:**

**If employee ID 101 appears three times, the result shows:**


| employee_id | total |
|---|---:|
| 101 | 3 |


**Q57. How do you find the top three salaries in each department?**

WITH ranked_employees AS (

    SELECT

        employee_id,

        name,

        department,

        salary,

        DENSE_RANK() OVER (

            PARTITION BY department

            ORDER BY salary DESC

        ) AS salary_rank

    FROM employees

)

SELECT *

FROM ranked_employees

WHERE salary_rank <= 3;


**Q58. How do you calculate the average salary for each department?**

SELECT

    department,

    ROUND(AVG(salary), 2) AS average_salary

FROM employees

GROUP BY department;

**Example:**

**If the IT department has salaries of ₹40,000 and ₹60,000, its average salary is ₹50,000.**

**Q59. How do you create and query a temporary view?**


**First, create the view:**

CREATE OR REPLACE TEMP VIEW high_salary_employees AS

SELECT *

FROM employees

WHERE salary > 50000;


**Then query it:**

SELECT *

FROM high_salary_employees;

**Example: If the table contains employees earning ₹40,000, ₹60,000, and ₹70,000, the view returns only the employees earning ₹60,000 and ₹70,000.**


**Q60. How do you update existing records and insert new records using MERGE?**


**Assume employees_target is the target table and employees_source contains incoming employee records.**


MERGE INTO employees_target t

USING employees_source s

ON t.employee_id = s.employee_id

WHEN MATCHED THEN

    UPDATE SET

        t.name = s.name,

        t.salary = s.salary

WHEN NOT MATCHED THEN

    INSERT (employee_id, name, salary)

    VALUES (s.employee_id, s.name, s.salary);


**Example:**

**The target has employee 101 with a salary of ₹40,000. The source has employee 101 with a salary of ₹50,000 and a new employee 102.**


## Another Scenario-Based Interview Questions:


### Q1. Your orders table contains 500 million records. Most queries filter by order date. What would you consider?

- I would evaluate date-based storage partitioning or the platform's supported data-layout optimizations. 
- Partition pruning can reduce unnecessary data scans. 
- I would also examine file sizes, query plans, and data distribution before deciding on the layout.

### Q2. Two large tables are frequently joined using customer_id. Would you always use bucketing?

- No. 
- I would first inspect the query plans, table formats, join strategy, data distribution, and platform support. 
- Bucketing can help in suitable environments, but it is not universally supported or beneficial.

### Q3. You need to share intermediate data between Spark sessions. Would you use a local temporary view?

- No, 
- not if the other session needs to access it. 
- I would consider a global temporary view within the same Spark application or a persistent table if the data must be shared beyond that scope.

### Q4. Your dashboard repeatedly calculates daily revenue from a huge transaction table. What would you consider?

- I would evaluate precomputed aggregates, a supported materialized view, or an incrementally maintained summary table. 
- The best solution depends on query frequency, freshness requirements, refresh cost, and platform capabilities.

### Q5. Your customer source contains both new customers and updates to existing customers. What would you use?

- I would consider a Delta Lake MERGE operation to update matching customers and insert new ones. 
- I would also validate source-key uniqueness and handle deletions or late-arriving updates if required.

### Q6. A pipeline accidentally overwrites correct data with incorrect values. What would you do?

- I would inspect the Delta table history and query an earlier version using time travel if the required version and data remain available. 
- After validating the correct state, I would determine whether a controlled restore or targeted correction is appropriate.

### Q7. Should you always cache a DataFrame used in a Spark job?

- No. 
- I would cache it when expensive computations are reused enough to justify the storage overhead. 
- For a one-time transformation, caching may provide no benefit.

### Q8. Is Spark SQL faster than the PySpark DataFrame API?

- Neither is inherently faster. 
- Equivalent operations can produce similar optimized execution plans. 
- I would compare the plans and runtime rather than selecting an API based on assumptions.
