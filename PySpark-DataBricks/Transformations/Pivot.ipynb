## pivot() :

- pivot() is used to `transforms distinct values from a column into separate columns`, allowing you to `summarize data by categories`.

- here we can use aggregation such as `sum(), count(), or avg()`.

### Before Pivoting:

| Employee | Month | Sales |
|----------|-------|-------|
| Arun     | Jan   | 10000 |
| Arun     | Feb   | 15000 |
| Bala     | Jan   | 12000 |
| Bala     | Feb   | 18000 |

### After Pivoting:

| Employee | Jan   | Feb   |
|----------|-------|-------|
| Arun     | 10000 | 15000 |
| Bala     | 12000 | 18000 |

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum

spark = SparkSession.builder.appName("PivotExample").getOrCreate()

sales_df = spark.createDataFrame([
    ("Arun", "Jan", 10000),
    ("Arun", "Feb", 15000),
    ("Bala", "Jan", 12000),
    ("Bala", "Feb", 18000),
    ("Arun", "Jan", 5000)
], ["EmployeeName", "Month", "Sales"])

pivot_df = (
    sales_df
    .groupBy("EmployeeName")
    .pivot("Month", ["Jan", "Feb"])
    .agg(sum("Sales"))
)

pivot_df.show()
```

### A retail company receives daily sales transactions from multiple branches. The manager wants a report comparing monthly revenue for each branch.

%md
### Input Data:

| Branch  | Month | Revenue |
|---------|-------|---------|
| Chennai | Jan   | 50000   |
| Chennai | Jan   | 30000   |
| Chennai | Feb   | 60000   |
| Madurai | Jan   | 40000   |
| Madurai | Feb   | 70000   |
| Madurai | Feb   | 10000   |


### Required Output: Monthly Revenue by Branch

| Branch  | Jan   | Feb   |
|---------|-------|-------|
| Chennai | 80000 | 60000 |
| Madurai | 40000 | 80000 |

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum

spark = SparkSession.builder.appName(
    "RetailSalesPivot"
).getOrCreate()

sales_df = spark.createDataFrame([
    ("Chennai", "Jan", 50000),
    ("Chennai", "Jan", 30000),
    ("Chennai", "Feb", 60000),
    ("Madurai", "Jan", 40000),
    ("Madurai", "Feb", 70000),
    ("Madurai", "Feb", 10000)
], ["Branch", "Month", "Revenue"])

print("Original sales data:")
sales_df.show()

# Convert month values into columns
monthly_report = (
    sales_df
    .groupBy("Branch")
    .pivot("Month", ["Jan", "Feb"])
    .agg(sum("Revenue"))
)

print("Monthly revenue report:")
monthly_report.show()
```


### Find the total revenue of each branch.

```python
branch_revenue = sales_df.groupBy("Branch").agg(
    sum("Revenue").alias("TotalRevenue")
)

branch_revenue.show()
```
