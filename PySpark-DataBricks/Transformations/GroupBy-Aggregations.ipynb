## groupBy() and Aggregations



`groupBy()` is used to group rows that have the same value in one or more columns.

`Aggregation` is used to calculate values such as total, average, count, minimum, and maximum, for each group.

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("GroupBy").getOrCreate()

employee_df=spark.createDataFrame([
    (101, "Arun", "IT", 40000),
    (102, "Bala", "HR", 30000),
    (103, "Kumar", "IT", 50000),
    (104, "Ravi",	"HR",	35000),
    (105, "Priya", "Sales",	45000)
],["EmployeeID","EmployeeName","Department","Salary"])

employee_df.show()
```

## groupBy() and Aggregations:

```python
from pyspark.sql.functions import avg

employee_df.groupBy("Department") \
    .agg(avg("Salary").alias("AverageSalary")) \
    .show()
```

```python
from pyspark.sql import SparkSession

from pyspark.sql.functions import (
    count, sum, avg, min, max,
    countDistinct, first, last,
    collect_list, collect_set
)

spark = SparkSession.builder.appName("GroupByAggregations").getOrCreate()

# Create Employee DataFrame
employee_df = spark.createDataFrame([
    (101, "Arun", "IT", 40000),
    (102, "Bala", "HR", 30000),
    (103, "Kumar", "IT", 50000),
    (104, "Ravi", "HR", 35000),
    (105, "Priya", "Sales", 45000)
], ["EmployeeID", "EmployeeName", "Department", "Salary"])

employee_df.show()

# Perform all aggregate functions
result_df = employee_df.groupBy("Department").agg(
    count("*").alias("TotalEmployees"),
    sum("Salary").alias("TotalSalary"),
    avg("Salary").alias("AverageSalary"),
    min("Salary").alias("MinimumSalary"),
    max("Salary").alias("MaximumSalary"),
    countDistinct("EmployeeID").alias("UniqueEmployees"),
    first("EmployeeName").alias("FirstEmployee"),
    last("EmployeeName").alias("LastEmployee"),
    collect_list("EmployeeName").alias("EmployeeList"),
    collect_set("EmployeeName").alias("UniqueEmployeeList")
)

result_df.show(truncate=False)
```


## Aggregate Functions in above example explained:

Aggregate functions is used to `perform calculations over group of rows and return a single result`.

| Function | Meaning | Example Output |
|---|---|---|
| `count("*")` | Counts the total number of rows. | `2` |
| `sum("Salary")` | Calculates the total salary. | `90000` |
| `avg("Salary")` | Calculates the average salary. | `45000` |
| `min("Salary")` | Finds the lowest salary. | `40000` |
| `max("Salary")` | Finds the highest salary. | `50000` |
| `countDistinct("EmployeeID")` | Counts unique employee IDs. | `2` |
| `first("EmployeeName")` | Returns the first value encountered in the group. | `Arun` |
| `last("EmployeeName")` | Returns the last value encountered in the group. | `Kumar` |
| `collect_list("EmployeeName")` | Collects values into an array, including duplicates. | `["Arun", "Kumar"]` |
| `collect_set("EmployeeName")` | Collects distinct values into an array. | `["Arun", "Kumar"]` |


## PySpark Aggregation Functions:

Aggregation functions perform calculations on multiple rows and return a single value or a result for each group.

| No. | Function | Definition |
|---:|---|---|
| 1 | `count()` | Counts the number of rows or non-null values in a column. |
| 2 | `sum()` | Calculates the total of values in a numeric column. |
| 3 | `avg()` | Calculates the average of values in a numeric column. |
| 4 | `mean()` | Calculates the average of values in a numeric column. |
| 5 | `min()` | Returns the minimum value in a column. |
| 6 | `max()` | Returns the maximum value in a column. |
| 7 | `first()` | Returns the first value encountered in a group; the result may be nondeterministic without a defined ordering strategy. |
| 8 | `last()` | Returns the last value encountered in a group; the result may be nondeterministic without a defined ordering strategy. |
| 9 | `countDistinct()` | Counts the number of distinct values in a column or expression. |
| 10 | `approx_count_distinct()` | Estimates the number of distinct values in a column. |
| 11 | `collect_list()` | Collects column values into an array, including duplicates. |
| 12 | `collect_set()` | Collects distinct column values into an array. |
| 13 | `stddev()` | Calculates the sample standard deviation of a column. |
| 14 | `stddev_pop()` | Calculates the population standard deviation of a column. |
| 15 | `variance()` | Calculates the sample variance of a column. |
| 16 | `var_pop()` | Calculates the population variance of a column. |
| 17 | `skewness()` | Measures the asymmetry of a column's value distribution. |
| 18 | `kurtosis()` | Measures the tail characteristics of a column's value distribution. |
| 19 | `corr()` | Calculates the Pearson correlation between two numeric columns. |
| 20 | `covar_samp()` | Calculates the sample covariance between two numeric columns. |
| 21 | `covar_pop()` | Calculates the population covariance between two numeric columns. |
| 22 | `percentile_approx()` | Calculates an approximate percentile of a column. |
| 23 | `median()` | Returns the median value of a column. |
| 24 | `sum_distinct()` | Calculates the sum of distinct values in a column. |
| 25 | `grouping()` | Indicates whether a column is aggregated away in a grouping set. |
| 26 | `grouping_id()` | Returns an integer bitmask identifying the grouping columns used in a grouping set. |

### Important Notes

- `avg()` and `mean()` both calculate the average.
- `stddev()` and `variance()` calculate sample statistics, while `stddev_pop()` and `var_pop()` calculate population statistics.
- `collect_list()` preserves duplicates, whereas `collect_set()` removes duplicates.
- `corr()` measures linear association; correlation does not establish causation.
- `grouping()` and `grouping_id()` are mainly used with advanced grouping operations such as `ROLLUP`, `CUBE`, and `GROUPING SETS`.
- `percentile_approx()` is useful for estimating percentiles efficiently on large datasets.
