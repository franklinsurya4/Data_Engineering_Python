
### Create Customers Dataset:

```python
customers_data = [
    (101, "Arun", "Chennai", "Gold"),
    (102, "Priya", "Bengaluru", "Silver"),
    (103, "Karthik", "Hyderabad", "Gold"),
    (104, "Divya", "Chennai", "Bronze"),
    (105, "Rahul", "Coimbatore", "Silver"),
    (106, "Meena", None, "Gold")
]

customers_df = spark.createDataFrame(
    customers_data,
    ["customer_id", "customer_name", "city", "membership"]
)

# customers_df.show()
display(customers_df)
```


## `show()` vs `display()`

| Feature | `show()` | `display()` |
|---|---|---|
| Purpose | Prints DataFrame | Visually renders DataFrame |
| Available in | Standard PySpark and Databricks | Databricks notebooks |
| Output | Text-based table | Interactive table |
| Charts | No | Yes, supports visualizations |
| Filtering/Sorting UI | No | Yes, provides UI options |
| Default rows | 20 | Displays data in the notebook UI, subject to display limits |


### Create Orders Dataset:

```python
orders_data = [
    (1001, 101, 301, "2026-01-05", "Delivered"),
    (1002, 102, 302, "2026-01-07", "Delivered"),
    (1003, 101, 301, "2026-01-10", "Delivered"),
    (1004, 103, 303, "2026-01-12", "Cancelled"),
    (1005, 104, 301, "2026-02-02", "Delivered"),
    (1006, 105, 304, "2026-02-10", "Delivered"),
    (1007, 102, 302, "2026-02-15", "Pending"),
    (1008, 106, 303, "2026-02-20", "Delivered"),
    (1009, 103, 303, "2026-03-01", "Delivered"),
    (1010, 101, 301, "2026-03-05", "Delivered"),
    (1011, 104, 301, "2026-03-10", "Delivered"),
    (1012, 105, 304, "2026-03-15", "Cancelled")
]

orders_df = spark.createDataFrame(
    orders_data,
    ["order_id", "customer_id", "store_id", "order_date", "order_status"]
).withColumn(
    "order_date", F.to_date("order_date")
)

display(orders_df)
```
