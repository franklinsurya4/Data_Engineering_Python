## 11. Important Concepts to Remember

| Concept | Interview Definition |
|---|---|
| Transformation | An operation that defines a new DataFrame or dataset, such as `filter()` or `select()`. |
| Action | An operation that triggers execution or returns a result, such as `show()` or `count()`. |
| Lazy evaluation | Spark generally delays executing transformations until an action or another execution-triggering operation requires the result. |
| Schema | Defines column names, data types, and structural information. |
| Null handling | Identifies, replaces, or removes missing values. |
| Data quality | Measures how accurate, complete, consistent, and valid the data is. |
| Shuffle | Redistributes data across partitions, often during joins and aggregations. |
| Partition | A portion of distributed data processed by Spark tasks. |
| Narrow transformation | A transformation that generally does not require a shuffle across partitions. |
| Wide transformation | A transformation that generally requires data redistribution across partitions. |

## 12. Common PySpark Imports

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.functions import *
from pyspark.sql.window import Window
from pyspark.sql.types import *
```

**Important notes:**

- `F.sum()` is a SQL aggregate expression, while `DataFrame.coalesce(n)` reduces the number of partitions. These are different APIs with different purposes.
- `groupBy()` is typically used for aggregation that reduces rows per group; window functions calculate results while preserving the original row-level structure.
- `intersect()` and `subtract()` remove duplicates, while `intersectAll()` and `exceptAll()` account for duplicate counts.
- `union()` matches columns by position, so ensure the column order and compatible data types are correct. Use `unionByName()` when matching by column names is appropriate.
- `sample()` and `sampleBy()` are probabilistic. They do not guarantee an exact row count.
- The statistical methods listed here are not all imported from `pyspark.sql.functions`; for example, `freqItems()` is accessed through `df.stat.freqItems()`.
