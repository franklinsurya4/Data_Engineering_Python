## Joins in Spark:

A Joins in Spark is an `operation` which is used to combines `rows from two DataFrames` based on join condition.

###  Types of joins:

1. Inner Join
2. Left Join
3. Right Join
4. Full Outer Join
5. Left Semi Join
6. Left Anti Join

8. Self Join


%md
# Types of Joins in PySpark

## 1. Inner Join
Returns the rows are matching in both DataFrames based on the join condition.

**Example:** Find employees who have matching sales records.

## 2. Left Outer Join (Left Join)
-  Returns all rows from the left DataFrame and matching rows from the right DataFrame. 
-  If no match exists, the right DataFrame columns contain NULL values.

**Example:** Find all employees and their sales details, including employees without sales.

## 3. Right Outer Join (Right Join)
- Returns all rows from the right DataFrame and matching rows from the left DataFrame. 
- If no match exists, the left DataFrame columns contain NULL values.

**Example:** Find all sales records, including records without matching employee details.

## 4. Full Outer Join
- Returns all rows from both DataFrames. 
- Matching rows are combined, and non-matching rows contain NULL values for the missing side.

**Example:** Find all employees and all sales records, including unmatched records from either DataFrame.

## 5. Left Semi Join
- Returns the rows from the left DataFrame that have at least one matching row in the right DataFrame. 
- Only left DataFrame columns are returned.

**Example:** Find employees who have at least one sales record.

## 6. Left Anti Join
-  Returns the rows from the left DataFrame that do not have any matching row in the right DataFrame. 
-  Only left DataFrame columns are returned.

**Example:** Find employees who do not have any sales records.

## 7. Cross Join
Returns the Cartesian product of two DataFrames, combining every row from the left DataFrame with every row from the right DataFrame.

**Example:** Generate every possible combination of employees and products.

## 8. Self Join
Joins a DataFrame with itself using aliases and a join condition.

**Example:** Find each employee's manager using the employee table.

## 2. Left Outer Join (Left Join):

- Returns all rows from the left DataFrame and matching rows from the right DataFrame.
- If no match exists, the right DataFrame columns contain NULL values.

Example: Find all employees and their sales details, including employees without sales.

```python
employees_df.join(sales_df, "EmployeeID", "left").show()
```

## 3. Right Outer Join (Right Join):

- Returns all rows from the right DataFrame and matching rows from the left DataFrame.
- If no match exists, the left DataFrame columns contain NULL values.

Example: Find all sales records, including records without matching employee details.

```python
employees_df.join(sales_df, "EmployeeID", "right").show()
```

## 4. Full Outer Join:

- Returns all rows from both DataFrames.
- Matching rows are combined, and non-matching rows contain NULL values for the missing side.

Example: Find all employees and all sales records, including unmatched records from either DataFrame.

```python
employees_df.join(sales_df, "EmployeeID", "full").show()
```

## 5. Left Semi Join:

- Returns the rows from the left DataFrame and that rows have at least one matching row in the right DataFrame.
- Only left DataFrame columns are returned.

```python
employees_df.join(sales_df, "EmployeeID", "left_semi").show()
```

## 6. Left Anti Join:

- Returns the rows from the left DataFrame and and that rows not have any matching row in the right DataFrame.
- Only left DataFrame columns are returned.

```python
employees_df.join(sales_df, "EmployeeID", "left_anti").show()
```

## 7. Cross Join:

Returns the Cartesian product of two DataFrames, combining every row from the left DataFrame with every row from the right DataFrame.

Example: Generate every possible combination of employees and products.

```python

from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("CrossJoin").getOrCreate()

employees = spark.createDataFrame([
    (1, "Arun"),
    (2, "Bala")
], ["employee_id", "name"])

products = spark.createDataFrame([
    (101, "Laptop"),
    (102, "Mouse")
], ["product_id", "product"])


result = employees.crossJoin(products)

result.show()
```


## 8. Self Join:

Joins a DataFrame with itself using aliases and a join condition.

Example: Find each employee's manager using the employee table.

```python

from pyspark.sql.functions import col

employees = spark.createDataFrame([
    (1, "Arun", None),
    (2, "Bala", 1),
    (3, "Chitra", 1),
    (4, "David", 2)
], ["employee_id", "name", "manager_id"])

emp = employees.alias("emp")
mgr = employees.alias("mgr")

result = emp.join(
    mgr,
    col("emp.manager_id") == col("mgr.employee_id"),
    "left"
).select(
    col("emp.employee_id"),
    col("emp.name").alias("employee_name"),
    col("mgr.name").alias("manager_name")
)

result.show()
```
