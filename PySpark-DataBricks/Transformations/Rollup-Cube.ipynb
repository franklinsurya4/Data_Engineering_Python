
## Rollup and Cube:

- **rollup() and cube()** are `aggregation methods used to calculate grand totals and subtotals` across multiple columns. 
- They are commonly used in `reports, business intelligence, and data warehousing`.



## rollup():

- **rollup()** is a `aggregation method`.
- **rollup()** is a `hierarchical aggregation used to calculating subtotals and grand total` on specified columns. 


**An e-commerce company wants to calculate sales revenue by Region and Product.**


## Create the DataFrame:

```python
from pyspark.sql import functions as F

sales_df = spark.createDataFrame([
    ("South", "Laptop", 50000),
    ("South", "Mobile", 20000),
    ("North", "Laptop", 60000),
    ("North", "Mobile", 30000)
], ["Region", "Product", "Revenue"])

sales_df.show()
```


## Apply rollup():

```python
rollup_df = (
    sales_df
    .rollup("Region", "Product")
    .agg(F.sum("Revenue").alias("TotalRevenue"))
)

rollup_df.orderBy("Region", "Product").show()
```


## How to understand the output:
- **North, Laptop** → `Laptop revenue` in the North region.
- **North, NULL** → `Total revenue` for the North region.
- **South, NULL** → `Total revenue` for the South region.
- **NULL, NULL** → `Grand total revenue` across all regions.


## cube():

- **cube()** is a `aggregation method`.
- **cube()** is a `multidimensional aggregation used to calculating subtotals and grand total` for all combinations specified group of columns.


**The company wants sales totals by region, by product, by region and product, and overall.**


## Apply cube()

```python
cube_df = (
    sales_df
    .cube("Region", "Product")
    .agg(F.sum("Revenue").alias("TotalRevenue"))
)

cube_df.orderBy("Region", "Product").show()
```


## Let's use Another real-time sales dataset

```python
sales_df2 = spark.createDataFrame([
    ("Chennai", "Electronics", 10000),
    ("Chennai", "Clothing", 5000),
    ("Madurai", "Electronics", 8000),
    ("Madurai", "Clothing", 4000)
], ["Branch", "Category", "Revenue"])

sales_df.show()
```


**A company wants to calculate:**

- **Revenue for each branch and category.**
- **Total revenue for each branch.**
- **Overall revenue across all branches.**


## rollup() aggregation:

```python
from pyspark.sql.functions import sum

sales_df2.rollup("Branch", "Category") \
    .agg(sum("Revenue").alias("TotalRevenue")) \
    .orderBy("Branch", "Category") \
    .show()
```


## Cube() aggregation:

```python
# Apply Cube aggregation
sales_df2.cube("Branch", "Category") \
    .agg(F.sum("Revenue").alias("TotalRevenue")) \
    .orderBy("Branch", "Category") \
    .show()
```
